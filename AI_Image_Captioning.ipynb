## 📚 Importing Required Modules

In this step, we import all the necessary libraries required for building the Image Captioning system using a pre-trained Generative AI model.

### 🔧 Modules Used:
- **Gradio** → For building an interactive web interface

- **PyTorch** → Backend framework for model execution
- **Warnings** → Suppress unnecessary warnings for cleaner output

> This setup prepares the environment for loading the model and processing input images efficiently.

```python
import gradio as gr
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import torch

import warnings
warnings.filterwarnings("ignore")
```

## 🤖 Loading Pretrained BLIP Model

In this step, we load a pre-trained image captioning model from Hugging Face.

### 🔍 Model Used:
- **BLIP (Bootstrapped Language Image Pretraining)**
- Variant: `blip-image-captioning-large`
- Provider: Salesforce

### ⚙️ Components:
- **Processor** → Handles image preprocessing (resizing, normalization, tensor conversion)
- **Model** → Generates captions from images using a vision-language architecture

> This model is capable of understanding visual content and generating meaningful natural language descriptions.

```python
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-large")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-large")
```

## 🧠 Generating Image Captions

In this step, we define a function that takes an input image and generates a descriptive caption using the pre-trained BLIP model.

### 🔄 Workflow:
1. Load and preprocess the image
2. Convert image into model-compatible tensor format
3. Generate caption using the model
4. Decode the output tokens into human-readable text

> This function serves as the core inference pipeline of the Image Captioning system.

```python
def generate_caption(image):
    raw_image = Image.open(image).convert("RGB")
    inputs = processor(raw_image, return_tensors  = "pt")
    with torch.no_grad():
        out = model.generate(**inputs)
    caption = processor.decode(out[0], skip_special_tokens = True)
    return caption
```

## 🌐 Building Interactive Web Interface with Gradio

In this step, we create a user-friendly web interface using Gradio to interact with the image captioning model.

### 🎯 Features:
- Upload an image from local system
- Generate caption using AI model
- Display output instantly

### ⚙️ Components:
- **Function (`fn`)** → Connects backend model
- **Input (Image)** → Accepts user image
- **Output (Textbox)** → Displays generated caption
- **Title & Description** → Improves UI/UX

> This transforms the model into an interactive web application without needing frontend development.

```python
iface = gr.Interface(
    fn = generate_caption,
    inputs = gr.Image(type = "filepath"),
    outputs = gr.Textbox(label = "Generated Caption"),
    title = "AI Image Captioning",
    description = "Welcome, Upload an Image and AI will generate a descriptive caption for it using Blip Model."
)
```

```python
if __name__ == "__main__":
    iface.launch(share=True)
```
